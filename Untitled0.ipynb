```python
import numpy as np
import matplotlib.pyplot as plt

# Create vectors
A = np.array([2, 4, 6])
B = np.array([1, 3, 5])

# Vector operations
print("Vector A:", A)
print("Vector B:", B)
print("Vector Addition:", A + B)
print("Vector Subtraction:", A - B)
print("Dot Product:", np.dot(A, B))

# Create matrices
M1 = np.array([[1, 2], [3, 4]])
M2 = np.array([[5, 6], [7, 8]])

print("\nMatrix M1:\n", M1)
print("Matrix M2:\n", M2)

# Matrix operations
print("\nMatrix Addition:\n", M1 + M2)
print("Matrix Multiplication:\n", np.dot(M1, M2))
print("Transpose of M1:\n", M1.T)
print("Inverse of M1:\n", np.linalg.inv(M1))

# Visualization
plt.figure(figsize=(8, 4))

plt.subplot(1, 2, 1)
plt.imshow(M1, cmap='Blues')
plt.title("Matrix M1")
plt.colorbar()

plt.subplot(1, 2, 2)
plt.imshow(M2, cmap='Greens')
plt.title("Matrix M2")
plt.colorbar()

plt.tight_layout()
plt.show()
```
