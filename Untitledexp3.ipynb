```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm, binom

# Normal distribution
x = np.linspace(-5, 5, 100)

mean = 0
std = 1

normal_values = norm.pdf(x, mean, std)

plt.figure(figsize=(8, 5))
plt.plot(x, normal_values, color='blue',
         label='Normal Distribution')

plt.title("Normal Probability Distribution")
plt.xlabel("Values")
plt.ylabel("Probability Density")
plt.legend()
plt.grid()
plt.show()

# Binomial distribution
n = 10
p = 0.5

x_binomial = np.arange(0, n + 1)
binomial_values = binom.pmf(x_binomial, n, p)

plt.bar(x_binomial, binomial_values, color='green')
plt.title("Binomial Distribution")
plt.xlabel("Number of Successes")
plt.ylabel("Probability")
plt.show()

# Bayes' Theorem
P_A = 0.01       # Probability of disease
P_B_given_A = 0.90
P_B_given_not_A = 0.05

P_not_A = 1 - P_A

P_B = (P_B_given_A * P_A) + \
      (P_B_given_not_A * P_not_A)

P_A_given_B = (P_B_given_A * P_A) / P_B

print("Bayes' Theorem Calculation")
print("P(A):", P_A)
print("P(B):", P_B)
print("P(A|B):", P_A_given_B)
print("Percentage:", P_A_given_B * 100)
```
