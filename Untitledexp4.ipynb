```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import entropy

# Actual probability distribution
P = np.array([0.7, 0.2, 0.1])

# Predicted probability distribution
Q = np.array([0.6, 0.3, 0.1])

# Calculate entropy
H_P = entropy(P, base=2)

# Calculate cross entropy
cross_entropy = -np.sum(P * np.log2(Q))

# Calculate KL divergence
KL_divergence = entropy(P, Q, base=2)

print("Actual Distribution:", P)
print("Predicted Distribution:", Q)

print("\nEntropy:", H_P)
print("Cross Entropy:", cross_entropy)
print("KL Divergence:", KL_divergence)

# Visualization
labels = ['Class 1', 'Class 2', 'Class 3']

x = np.arange(len(labels))
width = 0.35

plt.bar(x - width/2, P, width, label='Actual')
plt.bar(x + width/2, Q, width, label='Predicted')

plt.xlabel("Classes")
plt.ylabel("Probability")
plt.title("Actual vs Predicted Probability")
plt.xticks(x, labels)
plt.legend()
plt.grid(axis='y')

plt.show()

# Display comparison
metrics = [H_P, cross_entropy, KL_divergence]
names = ['Entropy', 'Cross Entropy', 'KL Divergence']

plt.bar(names, metrics, color=['blue', 'green', 'orange'])
plt.ylabel("Value (bits)")
plt.title("Information Theory Metrics")
plt.show()
```
